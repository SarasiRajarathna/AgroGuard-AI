# AgroGuard-AI: CropHelth & PlantVillage Exploratory Data Analysis & Transfer Learning

This notebook documents:
1. Ingestion of the `hansaka01/crophelth` and `mohanty/PlantVillage` datasets from Hugging Face.
2. Class, crop, and healthy vs. diseased statistical distributions.
3. Transfer learning using MobileNetV3 with class-weighted loss and cosine annealing.
4. Macro-F1 evaluation, per-class metrics, and confusion matrix analysis.

```python
import os
import json
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Load class mappings and stats
data_dir = Path("../data")
with open(data_dir / "dataset_report.json", "r") as f:
    report = json.load(f)

print(f"Total Images Analyzed: {report['total_images']:,}")
print(f"Total Pathology Classes: {report['num_classes']}")
print(f"Total Crops Covered: {report['num_crops']}")
```

```python
# Plot Healthy vs Diseased Distribution
labels = ['Healthy Foliage', 'Diseased Foliage']
counts = [report['distribution']['healthy_samples'], report['distribution']['diseased_samples']]
colors = ['#10b981', '#ef4444']

plt.figure(figsize=(6, 4))
plt.bar(labels, counts, color=colors, width=0.5)
plt.title("CropHelth Foliar Distribution (Healthy vs Diseased)")
plt.ylabel("Number of Images")
for i, v in enumerate(counts):
    plt.text(i, v + 2000, f"{v:,}", ha='center', fontweight='bold')
plt.ylim(0, 120000)
plt.show()
```

```python
# Load Evaluation Results
with open("../evaluation_report.json", "r") as f:
    eval_report = json.load(f)

metrics = eval_report['test_metrics']
print("=== MobileNetV3 Transfer Learning Evaluation on Test Set ===")
for k, v in metrics.items():
    print(f"{k:<20}: {v*100:.2f}%")
```
